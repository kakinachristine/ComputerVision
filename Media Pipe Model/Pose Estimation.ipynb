```python
# imports:
import numpy as np
import pandas as pd
import cv2
import mediapipe as mp
mp_drawing = mp.solutions.drawing_utils #Drawing the connecting lines
mp_pose = mp.solutions.pose #The AI model
```

# 1.Make Detection

```python
cap=cv2.VideoCapture(0)
#Set up mediapipe instance
with mp_pose.Pose(min_detection_confidence=0.5, min_tracking_confidence=0.5) as pose:
    while cap.isOpened():
        ret,frame=cap.read()

        #Recolor image to PGB
        image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        image.flags.writeable = False

        #Make Detection
        result=pose.process(image)

        #Recoloe back to BGR
        image.flags.writeable = True
        image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

        #Render Detection and draw them using draw_landmarks(requires two prams: image and the results from the model-process)
        mp_drawing.draw_landmarks(image,result.pose_landmarks,mp_pose.POSE_CONNECTIONS,
                                   #define the color of the joins, thickness and size of circle:
                                  mp_drawing.DrawingSpec(color=(245,117,66),thickness=2,circle_radius=2),
                                  mp_drawing.DrawingSpec(color=(245,66,230),thickness=2,circle_radius=2),
                                  )
        cv2.imshow('Mediapipe Feed',image)
        if cv2.waitKey(10) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()
```

# 3.CALCULATE ANGLES

```python
#Mathematical function to calculate an andle btwn three data points
def calculateAngle(a,b,c):
    a = np.array(a) #First
    b = np.array(b) #Mid
    c = np.array(c) #End

    radians = np.arctan2(c[1]-b[1],c[0]-b[0])- np.arctan2(a[1]-b[1],a[0]-b[0])
    angle = np.abs(radians* 180.0/np.pi)
    if angle > 180.0:
        angle = 360-angle

    return angle
```

# 4. CURL COUNTER

```python
cap=cv2.VideoCapture(0)

#Curl counter variables
counter = 0 # When exercising the counter starts at 0
stage= None # there are two stages when exercising UP AND DOWN

#Set up mediapipe instance
with mp_pose.Pose(min_detection_confidence=0.5, min_tracking_confidence=0.5) as pose:
    while cap.isOpened():
        ret,frame=cap.read()

        #Recolor image to PGB
        image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        image.flags.writeable = False

        #Make Detection
        result=pose.process(image)

        #Recoloe back to BGR
        image.flags.writeable = True
        image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

        #Extract the landmarks
        try:
            landmarks = result.pose_landmarks.landmark
            #Get coordinates

            shoulder =[landmarks[mp_pose.PoseLandmark.LEFT_SHOULDER.value].x,landmarks[mp_pose.PoseLandmark.LEFT_SHOULDER.value].y,]
            elbow =[landmarks[mp_pose.PoseLandmark.LEFT_ELBOW.value].x,landmarks[mp_pose.PoseLandmark.LEFT_ELBOW.value].y,]
            wrist =[landmarks[mp_pose.PoseLandmark.LEFT_WRIST.value].x,landmarks[mp_pose.PoseLandmark.LEFT_WRIST.value].y,]

            #Calculate the angle
            angle = calculateAngle(shoulder,elbow,wrist,)

            #Visualize the angle
            #putText- print text on img window and display angle value and location
            cv2.putText(image,str(angle),
                        tuple(np.multiply(elbow,[640,480]).astype(int)),
                        cv2.FONT_HERSHEY_SIMPLEX,0.5,(255,255,255),cv2.LINE_AA
                        )

            #CURL COUNTER LOGIC
            if angle > 160:
                stage ="down"
            if angle< 30 and stage == "down":
                stage = "up"
                counter += 1
                print(counter)

        except:

            pass

        # Render curl counter : prints the counter and stage values
        #set up status box
        cv2.rectangle(image,(0,0),(225,73),(245,117,16),-1)

        #Rep Data
        cv2.putText(image,'REPS',(15,12),
                    cv2.FONT_HERSHEY_SIMPLEX,0.5,(0,0,0),1,cv2.LINE_AA)
        cv2.putText(image,str(counter),(10,60),
                    cv2.FONT_HERSHEY_SIMPLEX,2,(255,255,255),2,cv2.LINE_AA)

        #stage data
        cv2.putText(image,"STAGE",(65,12),
                    cv2.FONT_HERSHEY_SIMPLEX,0.5,(0,0,0),1,cv2.LINE_AA)
        cv2.putText(image,stage,(60,60),
                    cv2.FONT_HERSHEY_SIMPLEX,2,(255,255,255),2,cv2.LINE_AA)


        #Render Detection and draw them using draw_landmarks(requires two prams: image and the results from the model-process)
        mp_drawing.draw_landmarks(image,result.pose_landmarks,mp_pose.POSE_CONNECTIONS,
                                   #define the color of the joins, thickness and size of circle:
                                  mp_drawing.DrawingSpec(color=(245,117,66),thickness=2,circle_radius=2),
                                  mp_drawing.DrawingSpec(color=(245,66,230),thickness=2,circle_radius=2),
                                  )
        cv2.imshow('Mediapipe Feed',image)
        if cv2.waitKey(10) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()
```
